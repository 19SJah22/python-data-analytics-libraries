```python
# ===============================
# 00_all_libraries.ipynb
# ===============================

# -----------------------------
# 1️⃣ Import libraries
# -----------------------------
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.datasets import load_iris
from sklearn.metrics import mean_squared_error, accuracy_score

# -----------------------------
# 2️⃣ Pandas Example
# -----------------------------
print("=== Pandas Example ===")
df = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv")
print(df.head())
print("Average age by sex:\n", df.groupby("sex")["age"].mean())

# -----------------------------
# 3️⃣ NumPy Example
# -----------------------------
print("\n=== NumPy Example ===")
arr = np.array([10, 20, 30, 40, 50])
print("Array:", arr)
print("Mean:", np.mean(arr))
print("Standard Deviation:", np.std(arr))
matrix = np.arange(1, 10).reshape(3, 3)
print("3x3 Matrix:\n", matrix)

# -----------------------------
# 4️⃣ Matplotlib Example
# -----------------------------
print("\n=== Matplotlib Example ===")
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
plt.plot(x, y, marker='o', color='b')
plt.title("Simple Line Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.show()

# -----------------------------
# 5️⃣ Seaborn Example
# -----------------------------
print("\n=== Seaborn Example ===")
tips = sns.load_dataset("tips")
sns.scatterplot(x="total_bill", y="tip", data=tips)
plt.title("Bill vs Tip (Seaborn)")
plt.show()

sns.boxplot(x="day", y="total_bill", data=tips)
plt.title("Total Bill by Day (Seaborn)")
plt.show()

# -----------------------------
# 6️⃣ Scikit-learn Example
# -----------------------------
print("\n=== Scikit-learn Example ===")
iris = load_iris()
X, y = iris.data, iris.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
model = LinearRegression()
model.fit(X_train, y_train)
predictions = model.predict(X_test)
print("Sample predictions:", predictions[:5])
print("Mean Squared Error:", mean_squared_error(y_test, predictions))
```
